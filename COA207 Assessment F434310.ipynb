```python
graph = {
    'Colosseum': {'Palatine Hill': 10, 'Roman Forum': 20},
    'Roman Forum': {'Colosseum': 20, 'Trevi Fountain': 30, 'Torre Argentina': 50},
    'Palatine Hill': {'Colosseum': 10, 'Torre Argentina': 20},
    'Trevi Fountain': {'Roman Forum': 30, 'Via Del Corso': 10},
    'Pantheon': {'Trevi Fountain': 40, 'Piazza Navona': 10, 'Piazza Del Popolo': 50},
    'Piazza Navona': {'Campo De Fiori': 50, 'Castel Sant Angelo': 30, 'Pantheon': 10},
    'Campo De Fiori': {'Torre Argentina': 50, 'Piazza Navona': 50},
    'Piazza Del Popolo': {'Via Del Corso': 50, 'Pantheon': 50, 'Prati District': 10},
    'Via Del Corso': {'Piazza Del Popolo': 50},
    'Torre Argentina': {'Palatine Hill': 20, 'Campo De Fiori': 50, 'Roman Forum': 50, 'Trastevere': 10},
    'Trastevere': {'Torre Argentina': 10, 'Castel Sant Angelo': 50},
    'Castel Sant Angelo': {'Trastevere': 50, 'Piazza Navona': 30, 'Vatican City': 10},
    'Prati District': {'Piazza Del Popolo': 10, 'Vatican City': 10},
    'Vatican City': {'Castel Sant Angelo': 10, 'Prati District': 10}
}


ztl_zones = {'Colosseum', 'Roman Forum', 'Piazza Navona', 'Trevi Fountain'}
traffic_zones = {'Trastevere', 'Piazza Del Popolo', 'Vatican City', 'Campo De Fiori', 'Palatine Hill'}

def node_penalty(node):
  if node in ztl_zones:
    return 15
  elif node in traffic_zones:
    return 10
  return 0

nodes = ', '.join(graph.keys())
print(f"Available locations: {nodes}\n")

start = input("Enter start location: ").title()
end = input("Enter end location: ").title()

if start not in graph:
    print(f"'{start}' is not a valid location.")
elif end not in graph:
  print(f"'{end}' is not a valid location.")
elif start == end:
  print("No route: start and end location are the same.")
else:
    distances = {node: float('inf') for node in graph}
    base_distances = {node: float('inf') for node in graph}
    distances[start] = 0
    base_distances[start] = 0
    unvisited = set(graph.keys())
    previous_nodes = {node: None for node in graph}

    while unvisited:
       current = min(unvisited, key=lambda node: distances[node])

       if distances[current] == float('inf'):
            break

       unvisited.remove(current)

       for neighbor, weight in graph[current].items():
            distance = distances[current] + weight + node_penalty(neighbor)
            if distance < distances[neighbor]:
              distances[neighbor] = distance
              base_distances[neighbor] = base_distances[current] + weight
              previous_nodes[neighbor] = current

    if distances[end] == float('inf'):
        print(f"No path from {start} to {end}.")
    else:
        path = []
        current_node = end

        while current_node is not None:
          path.append(current_node)
          current_node = previous_nodes[current_node]

        path.reverse()

        ztl_nodes = [node for node in path if node in ztl_zones]
        traffic_nodes = [node for node in path if node in traffic_zones]
        ztl_penalty = len(ztl_nodes) * 15
        traffic_penalty = len(traffic_nodes) * 10
        base_time = base_distances[end]
        total_time = base_time + ztl_penalty + traffic_penalty

        print(f"\nShortest route from {start} to {end}:")
        print(" -> ".join(path))
        print(f"Base time: {base_time} minutes")
        if ztl_penalty:
          print(f"ZTL time waste due to ({', '.join(ztl_nodes)}): +{ztl_penalty} minutes")
        if traffic_penalty:
            print(f"Traffic time waste due to ({', '.join(traffic_nodes)}): +{traffic_penalty} minutes")
        print(f"Total time cost: {total_time} minutes")
```
